### Step 0 — Make some example data
If you don't have an XRD file yet, run this cell. It creates a practice file called **xrd.csv** with two columns: `two_theta` and `intensity`.

*This is simulated data for learning, not a real measurement.*

```python
import numpy as np, pandas as pd
rng=np.random.default_rng(1)
tt=np.round(np.arange(20,80.001,0.02),2)
# Simulated spinel-ferrite-like pattern (approximate peak positions, Cu K-alpha)
# This is EXAMPLE data for practice, not a real measurement.
peaks=[(18.3,0.12),(30.1,0.35),(35.5,1.0),(37.1,0.08),(43.1,0.22),(53.5,0.10),(57.0,0.30),(62.6,0.40),(74.0,0.08)]
y=np.zeros_like(tt)
for c,h in peaks:
    w=0.25
    y+=h*(0.6*np.exp(-4*np.log(2)*(tt-c)**2/w**2)+0.4/(1+4*(tt-c)**2/w**2))
bg=0.06+0.05*np.exp(-(tt-20)/15)
I=(y+bg)*1000+rng.normal(0,8,len(tt))
pd.DataFrame({'two_theta':tt,'intensity':np.round(I,1)}).to_csv('xrd.csv',index=False)
print('Created xrd.csv with', len(tt), 'data points')
```

### Step 1 — The 6 lines ✨

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("xrd.csv")
plt.plot(df["two_theta"], df["intensity"])
plt.xlabel("2θ (degrees)"); plt.ylabel("Intensity (a.u.)")
plt.savefig("xrd.png", dpi=300)
plt.show()   # shows the plot in the notebook
```

### What each line does
| Line | Meaning |
|---|---|
| `import pandas as pd` | loads the tool that reads tables (like Excel) |
| `import matplotlib.pyplot as plt` | loads the plotting tool |
| `pd.read_csv("xrd.csv")` | reads your file into a table called `df` |
| `plt.plot(x, y)` | draws intensity against 2θ |
| `plt.xlabel / plt.ylabel` | writes the axis labels |
| `plt.savefig(..., dpi=300)` | saves a high-quality image for your paper |

To download the image in Colab: click the 📁 folder icon on the left → right-click **xrd.png** → Download.

### Bonus — a cleaner, journal-style figure

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(df["two_theta"], df["intensity"], color="black", linewidth=1)
ax.set_xlabel("2θ (degrees)", fontsize=13)
ax.set_ylabel("Intensity (a.u.)", fontsize=13)
ax.set_xlim(20, 80)
ax.set_yticks([])                        # XRD intensity is in arbitrary units
for side in ["top", "right"]:
    ax.spines[side].set_visible(False)
plt.tight_layout()
plt.savefig("xrd_clean.png", dpi=300)
plt.show()
```
